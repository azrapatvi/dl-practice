```python
from tensorflow.keras.datasets import imdb

(X_train,y_train),(X_test,y_test)=imdb.load_data()
```

```python
X_train
```

```python
# so the data is already integer encodedd we dont need too do tokenization or tokenizer.fit_on_text or tokenizer.texts_To_Sequences

len(X_train[0])
```

```python
len(X_train[1])
```

```python
# so as you can see the lenght of sentences is not the same so we have to doo padding

from tensorflow.keras.utils import pad_sequences

X_train=pad_sequences(X_train,padding='post')
X_test=pad_sequences(X_test,padding='post')
```

```python
# create rnn model

from tensorflow.keras import Sequential
from tensorflow.keras.layers import SimpleRNN,Dense

rnn=Sequential([

    SimpleRNN(72,input_shape=(2494,1),return_sequences=False),
    Dense(1,activation='sigmoid')
])
```

```python
rnn.summary()
```

```python
rnn.compile(optimizer='adam',
            loss='binary_crossentropy',
            metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop=EarlyStopping(
    monitor='val_accuracy',
    mode='max',
    restore_best_weights=True,
    patience=5
)

rnn_history=rnn.fit(X_train,y_train,epochs=20,verbose=1,validation_data=(X_test,y_test),callbacks=[early_stop])
```
