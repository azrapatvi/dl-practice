```python
import numpy as np

sentences = [
    "This product exceeded all my expectations, absolutely love it!",
    "The customer service team was incredibly helpful and quick to respond.",
    "Best movie I've watched this year, the acting was phenomenal.",
    "I'm so happy with how the project turned out.",
    "The food at this restaurant was delicious and beautifully presented.",
    "Great job on the presentation, really impressed everyone in the room.",
    "This app has made my daily routine so much easier.",
    "Highly recommend this course, learned a lot in a short time.",
    "The product broke within two days, totally disappointed.",
    "Customer support was rude and unhelpful throughout the call.",
    "Worst experience I've had at a restaurant, the food was cold and bland.",
    "The movie was a waste of time, poorly written and boring.",
    "I regret buying this, it doesn't work as advertised.",
    "The delivery was late and the packaging was damaged.",
    "This app keeps crashing and it's extremely frustrating to use.",
    "Terrible service, I won't be coming back here again.",
    "The package arrived on Tuesday as scheduled.",
    "The meeting has been rescheduled to next Monday.",
    "This laptop has a 15-inch screen and 8GB RAM.",
    "The store opens at 9 AM and closes at 8 PM.",
    "He mentioned that the report will be ready by Friday.",
    "The train departs from platform 3.",
    "The document contains five sections and an appendix.",
    "The weather today is partly cloudy with mild temperatures."
]
```

```python
# there is a tokenizer in keras to tokenize the sentences that we have
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer=Tokenizer(oov_token='<OOV>') # means if you get out of vocabulary words
```

```python
tokenizer.fit_on_texts(sentences)
```

```python
tokenizer.word_index # will show you for which word what indexis assigned
```

```python
tokenizer.word_counts # will give you frequency of the words
```

```python
sequences=tokenizer.texts_to_sequences(sentences)

sequences
```

```python
# now ass thee len of each sentences is diffenret so for that to make evry lenght of same size we will do padding

from keras.utils import pad_sequences

sequences=pad_sequences(sequences,padding='post') # means add zzero afterr the sentences
```
