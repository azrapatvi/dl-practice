```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[ "the glass of milk",
"the glass of juice",
"the cup of tea",
"I am a good boy",
"I am a good developer",
"understand the meaning of words",
"your videos are good"]
```

```python
sent
```

```python
# define vocabulary size
voc_size=10000
```

```python
for words in sent:
  print(words)
```

```python
#one hot representation

one_hot_representation=[one_hot(words,voc_size) for words in sent]
one_hot_representation
```

```python
# it means at the index 845 only i have the word "the" remianing else numbers are 0 ->sparse matrix

from tensorflow.keras import Sequential
from tensorflow.keras.utils import pad_sequences

import numpy as np

```

```python
sent_len=8
embedded_docs=pad_sequences(one_hot_representation,maxlen=sent_len,padding='pre')
```

```python
embedded_docs
```

```python
# feature representation - 300 dimensions means represnt my each word with 300 dimensions

from tensorflow.keras.layers import Dense,Embedding

dim=10 # so now after thiss every word is then represented by 10 dimensions

model=Sequential()
model.add(
    Embedding(
        input_dim=voc_size,
        output_dim=dim,
        input_shape=(sent_len,)
    )
)
model.compile(optimizer='adam')
```

```python
model.summary()
```

```python
embedded_docs
```

```python
model.predict(embedded_docs)
```
