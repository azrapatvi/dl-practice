```python
from tensorflow.keras.datasets import imdb

max_features = 10000   # vocabulary size

(X_train,y_train),(X_test,y_test)=imdb.load_data(num_words=max_features) #it means i want vocabulary size 10000 only the vocabularyy size is so bing but i want only of 10000
```

```python
# padding becuasee currently the data is not of same lenght

from tensorflow.keras.utils import pad_sequences

max_len=500

X_train=pad_sequences(X_train,maxlen=max_len)
X_test=pad_sequences(X_test,maxlen=max_len)
```

```python
len(X_train[0])
```

```python
# train rnn

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Embedding,SimpleRNN

model=Sequential()

model.add(Embedding(input_dim=max_features,output_dim=128,input_shape=(max_len,))) # this will take the each word of 500 dimenesions and convert it into vecotr of 128 ...
model.add(SimpleRNN(128,activation='relu'))
model.add(Dense(1,activation='sigmoid'))

```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop=EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history=model.fit(X_train,y_train,validation_data=(X_test,y_test),epochs=50,verbose=1,callbacks=[early_stop])
```

```python
loss,acc=model.evaluate(X_test,y_test)

print(loss)
print(acc)
```

```python
# prediction

def preprocess_text(text):
  text = text.lower()
  words = text.split()
  text_index = [word_index.get(word, 2) + 3 for word in words]
  padded_text = pad_sequences([text_index], maxlen=max_len)
  return padded_text
```

```python
new_text="this movie was terrible and a complete waste of time"

padded_new_text=preprocess_text(new_text)

padded_new_text
```
